## Loading the adult census dataset

````python
# Loading adult census dataset
import pandas as pd

adult_census = pd.read_csv("datasets/adult-census-numeric.csv")
````

````python
# Let's take a look at the data
adult_census
````

**The target in this dataset is the 'class' column.**

````python
# Set the target 
target_name = 'class'
target = adult_census[target_name]
target
````

## Separate the data and the target

````python
# Drop the target column
data = adult_census.drop(columns=[target_name])
data
````

````python
# Separate the non-numerical features
non_numerical_features = list()
for column in adult_census.columns:
    if adult_census[column].dtype == 'object' and column != target_name:
        non_numerical_features.append(column)
````

````python
# Drop the non-numerical features
data = data.drop(columns=non_numerical_features)
data
````

````python
# Let's see the data
data.columns
````

````python
print(
    f"The data contains {data.shape[0]} samples and "
    f"{data.shape[1]} features"
)
````

## Fit a model and make predictions

We will use _K-nearest neighbors algorithm_ strategy. To predict the target of a new sample, a k-nearest neighbors takes into account its **k** closest samples in the training set and predicts the mojority of these samples.

````python
# Training the model
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier()
_ = model.fit(data, target)
````

In sckit-learn an object that has ```fit``` method is called an **estimator**. If the estimator additionally has:
- a ```predict_ method```, it is called a **predictor**. Examples of predictores are classifiers or regressors.
- a ```transform``` method, it is called **transformer**. Examples of transformers are scalers or encoders. We will see more about transformers in the next notebook.

The method ```fit``` is composed of two elements: (i) a **learn algorithm** and (ii) some **model states**. The learning algorithm takes the training data and training target as input and sets the model states. These model states are later used to either predict or transform data as explained above.

````python
# Let's use our model to make some predictions using the same dataset
target_predicted = model.predict(data)
````

````python
# See the firsts 5 predicted targets
target_predicted[:5]
````

````python
# Let's compare with the true targets
target[:5]
````

````python
# Even we can check if the predictions are correct
target_predicted[:5] == target[:5]
````

````python
print(
    "Number of correct prediction: "
    f"{(target[:5] == target_predicted[:5]).sum()} / 5"
)
````

````python
# We can also calculate the average of the sucess rate
(target == target_predicted).mean()
````

## Train-test data split

### Because the original dataset is not available, we'll split the current one

````python
from sklearn.model_selection import train_test_split

data_train, data_test, target_train, target_test = train_test_split(
    data,
    target,
    test_size=0.2,
    random_state=37
)
````

````python
_ = model.fit(data_train, target_train)
````

````python
# Let's calculate the accuracy of our model
accuracy = model.score(data_test, target_test)
model_name = model.__class__.__name__

print(f"The test accuracy of {model_name} is {accuracy:.3f}")
````
